# Training

In the learn method is called the train method.

Parameters: 
- ep_len_mean: average episode length
- ep_rew_mean: average reward per episode
- fps: frames per second
- iterations: how many update iterations have happeened so far
- time_elapsed: time since training started
- total_timesteps: cumulative environmente steps taken

- approx_kl: approximate KL divergence between new and old policy --> measured how much policy changed
- clip_fraction: fraction of updates where the policy change was clipped --> how often the policy tried to change too much
- clip_range: epsilon value --> the allowed policy change range
- entropy_loss: policy entropy 
- explained_variance: measures how well the value function explains the returns 
- learning rate
- loss: total training loss
- n_updates: number of updates
- policy_gradient_loss: actual loss used to update the policy
- value_loss: loss from value function --> critic

```python
model.learn(total_timesteps=25000, progress_bar=True)
```

# Inference Loop

```python
env.env.env.env.render_mode = "human"
```

```python

rewards = []
episode_rewards = []
# Reset the environment to generate the first observation
observation, info = env.reset(seed=42)
for _ in range(1000):
    # this is where you would insert your policy
    action, state = model.predict(observation, deterministic=True)
    # step (transition) through the environment with the action
    # receiving the next observation, reward and if the episode has terminated or truncated
    observation, reward, terminated, truncated, info = env.step(action)
    episode_rewards.append(reward)
    print(len(episode_rewards))

    # If the episode has ended then we can reset to start a new episode
    if terminated or truncated:
        print(f"Episode ended with reward {sum(episode_rewards)}")
        observation, info = env.reset()
        rewards.append(episode_rewards)
        episode_rewards = []
print(f"Total rewards: {sum([sum(ep_rewards) for ep_rewards in rewards])}")


```

```python
env.close()
pygame.quit()
```
