# Imports

```python
import gymnasium as gym
import sys
import mlflow
from typing import Any, Dict, Tuple, Union
import numpy as np
from stable_baselines3 import PPO
from stable_baselines3.common.logger import HumanOutputFormat, Logger, KVWriter
import pygame
import torch
```

# MLFlow

```python
class MLflowOutputFormat(KVWriter):
    """
    Dumps key/value pairs into MLflow's numeric format.
    """

    def write(
        self,
        key_values: Dict[str, Any],
        key_excluded: Dict[str, Union[str, Tuple[str, ...]]],
        step: int = 0,
    ) -> None:
        for (key, value), (_, excluded) in zip(
            sorted(key_values.items()), sorted(key_excluded.items()), strict=True
        ):
            if excluded is not None and "mlflow" in excluded:
                continue

            if isinstance(value, np.ScalarType) and not isinstance(value, str):
                mlflow.log_metric(key, value, step)
```

```python
loggers = Logger(
    folder=None,
    output_formats=[HumanOutputFormat(sys.stdout), MLflowOutputFormat()],
)
```

# Training

```python
# Initialise the environment
env = gym.make("CartPole-v1", render_mode=None)

# Train the model
mlflow.set_experiment("PPO-CartPole-v1") # Set the experiment name
with mlflow.start_run(run_name="PPO-CartPole-v1"):
    mlflow.log_params({"env": "CartPole-v1", "model": "PPO"})
    mlflow.log_metric("training_iteration", 1)
    model = PPO("MlpPolicy", env, verbose=1)
    model.set_logger(loggers)
    model.learn(total_timesteps=10000, log_interval=1, progress_bar=True)

    with open("output.txt", "w") as f:
        f.write("Model results")
    mlflow.log_artifact("output.txt")
print("PPO model")
```
